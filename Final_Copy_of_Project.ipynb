```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
```

```python
df=pd.read_csv('Ardent project 22.csv')
df.shape
```

```python
df.head()
```

```python
#Feature selection
x=df.iloc[:,[3,4]].values
print(x)
```

```python
kmeans=KMeans(n_clusters=5)
kmeans.fit(x)
print(kmeans.cluster_centers_)

```

```python
df.info()
```

```python
y=print(kmeans.labels_)
```

```python
plt.scatter(x[:,0],x[:,1],c=kmeans.labels_,cmap='Greens')
plt.show()
```

```python
kmeans.predict([[55,52]])
```

```python

plt.scatter(x[:,0], x[:,1], c= kmeans.labels_, s = 50)
plt.scatter(kmeans.cluster_centers_[:,0],kmeans.cluster_centers_[:,1], c = 'r', s = 100)
plt.grid()
plt.title('Annual Income VS Spending Score')
plt.xlabel('Annual Income')
plt.ylabel('Spending Score')
plt.show()
```

```python
# Model prediction

m=int(input("Enter the Annual income of the customer: "))
n = int(input("Enter spending score of the customer: "))
r = kmeans.predict([[m,n]])
print("The customer has been successfully classified into segment  : ", r)
```
